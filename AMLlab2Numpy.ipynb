```python
#Given:
#salary = np.array([25000, 32000, 28000, 45000,
#                 50000, 38000, 42000, 30000])
salary=np.array([25000, 32000, 28000, 45000, 50000, 38000, 42000, 30000])
#Find:

#Total salary
print("Total salary : ",salary.sum())
#Average salary
print("Average salary : ",salary.mean())
#Highest salary
print("Highest salary : ",salary.max())
#Lowest salary
print("Lowest salary : ",salary.min())
#Median salary
print("Median salary : ",np.median(salary))
#Standard deviation
print("Standard deviation : ",np.std(salary))
```

```python
#Given: marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])
# Using NumPy boolean indexing, find:

marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

# Marks greater than 75
cond1=marks>75
result=marks[cond1]
print(result)

# Marks between 50 and 80
cond2=(marks>=50) & (marks<=80)
result=marks[cond2]
print(result)

# Number of students who scored below 50
cond3=marks<50
result=marks[cond3]
print(result.size)

#Percentage of students who scored 60 or above
cond4=marks>60
result=marks[cond4]
print((result.size)/(marks.size)*100)

```

```python
#10. The following array contains marks of four students in three subjects:

# marks = np.array([
#   [80, 75, 90],
#   [65, 70, 72],
#    [88, 92, 85],
#    [55, 60, 58]
# ])

marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])
total_marks=np.sum(marks,axis=1)
print("Total marks of each student is :",total_marks)

average1 = np.mean(marks, axis=1)
print("Average marks of each student is :",average1)

average2 = np.mean(marks, axis=0)
print("Average marks in each subject :",average2)

print("Student with the highest total is :",total_marks.argmax()+1)

print("Highest mark in each subject is :",marks.max(axis=0))
```

```python
# Daily temperatures for two weeks are:

temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])
print("Mean temperature : ",temperature.mean())
print("Maximum and minimum temperatures : "  ,temperature.max(),"and",temperature.min())
avg=np.mean(temperature)
result=temperature[temperature>avg]
print("Number of days above the average temperature : ",len(result))
condition=temperature>34
print("Temperature greater than 34 : ",temperature[condition])
diff=temperature.max()-temperature.min()
print("Difference between maximum and minimum temperature : ",diff)
std=np.std(temperature)
print("Standard deviation : ",std)
```

```python
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])
print("Total annual sales : ",sales.sum())
print("Average monthly sales : ",sales.mean())
# Month with the highest sales
print("Month with the highest sales : ",sales.argmax()+1)
print("Month with the lowest sales : ",sales.argmin()+1)
condition2=sales>20000
print("Months with sales greater than 20000 : ",len(sales[condition2]))
# Percentage increase from January to December
increase=(sales[-1]-sales[0])/sales[0]*100
print("Percentage increase from January to December : ",increase)
```
