```python
def fib(n):
  a=0;b=1
  print(a,b)
  for i in range(n):
    a,b=b,a+b
    print(b)

fib(10)
```

```python
li=[8,4,6,5,3]
n=len(li)
for i in range(n):
  for j in range(n-i-1):
    if li[j]>li[j+1]:
      li[j],li[j+1] = li[j+1],li[j]
print(li)
```
