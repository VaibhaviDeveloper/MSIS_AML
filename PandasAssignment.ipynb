```python
import pandas as pd
```

```python
# 1. Create a Pandas Series containing the marks: 78, 85, 92, 67, 88
 # Display:
 #Complete Series
marks=pd.Series([78, 85, 92, 67, 88])
print(marks)
# Values
print(marks.values)
# Index
print(marks.index)
# Data type
print(marks.dtype)
# Mean
print(marks.mean())
# Maximum
print(marks.max())
# Minimum
print(marks.min())
```

```python
# 2.Create a Series containing:

# | Student | Marks |
# | ------- | ----: |
# | Asha    |    85 |
#	| Ravi    |    78 |
#	| Meena   |    92 |
#	| Kiran   |    74 |
#	| Arun    |    88 |
#	Display the marks of Meena.
students=pd.Series([85,78,92,74,88],index=['Asha','Ravi','Meena','Kiran','Arun'])
print(students)
```

```python
print(students["Meena"])
```

```python
# 3. Create the following DataFrame:

# | Name  | Age | Department | Marks |
# | ----- | --: | ---------- | ----: |
# | Asha  |  22 | CSE        |    85 |
# | Ravi  |  23 | ECE        |    78 |
# | Meena |  21 | CSE        |    92 |
# | Kiran |  24 | ME         |    74 |
# | Arun  |  22 | ECE        |    88 |
Student={
    "Name":["Asha","Ravi","Meena","Kiran","Arun"],
    "Age":[22,23,21,24,22],
    "Department":["CSE","ECE","CSE","ME","ECE"],
    "Marks":[85,78,92,74,88]
}
# Display:Complete DataFrame
df=pd.DataFrame(Student)
print(df)
```

```python
# First 3 rows
print(df.head(3))
```

```python
# Last two rows
print(df.tail(2))
```

```python
# Shape
print(df.shape)
```

```python
# Column names
print(df.columns)
```

```python
# Data Types
print(df.dtypes)
```

```python
# 4. Using the previous DataFrame, display:

#	info()
print(df.info())
print("************************************")
#	describe()
print(df.describe())
print("************************************")
#	Number of rows
print(len(df))
print("************************************")
#	Number of columns
print(len(df.columns))
```

```python
# 5. Using df, display:

# Name column
print(df["Name"])
print("************************************")
# Marks Column
print(df["Marks"])
print("************************************")
# Name and marks together
print(df[["Name","Marks"]])
```

```python
# Display First row using iloc
print(df.iloc[:1,:])
print("***************************")
# Third row using iloc
print(df.iloc[3:4,:])
print("****************************")
#Rows 1 to 3
print(df.iloc[1:4,:])
print(df.loc[:,["Name","Marks"]])
```

```python
# 7. Find:
# Students scoring above 80.
print(df.loc[df["Marks"]>80])
print("**********************************************")
# Students scoring between 75 and 90
print(df.loc[(df["Marks"]>75) & (df["Marks"] < 90)])
print("**********************************************")
# Students belonging to CSE.
print(df.loc[df["Department"]=="CSE"])
print("**********************************************")
# Student belonging to CSE and marks above 80
print(df.loc[(df["Department"]=="CSE") & (df["Marks"]>80)])
```

```python
print(df.loc[(df["Department"]=="CSE") | (df["Department"]=="ECE") ])
```

```python
# 9. Add a new column called Result. Students scoring 75 or above should receive "Pass", otherwise "Fail".
df["Result"]=["Pass" if marks>=75 else "Fail" for marks in df["Marks"]]
print(df)
```
